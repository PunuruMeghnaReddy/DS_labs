```python

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_iris
iris=load_iris()
df=pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species']=iris.target
plt.figure(figsize=(6,4))
sns.histplot(df['sepal length (cm)'], bins=20, kde=True,color='orange')
plt.title("Histogram of Sepal Length")
plt.show()

#boxplot
plt.figure(figsize=(6,4))
sns.boxplot(x=df['sepal length (cm)'])
plt.title("Boxplot of Sepal Length")
plt.show()

#pie-chart
species_counts = df['species'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(species_counts,labels=iris.target_names,autopct='%1.1f%%')
plt.title("Pie Chart of Species Distribution")
plt.show()
```

```python
#scatter plot
plt.figure(figsize=(6,4))
sns.scatterplot(x=df['sepal length (cm)'], y=df['sepal width (cm)'], hue=df['species'], palette='Set1')
plt.title("Scatter Plot of Sepal Length vs Sepal Width")
plt.show()

#line-chart
plt.figure(figsize=(6,4))
sns.lineplot(x=df['sepal length (cm)'], y=df['sepal width (cm)'],hue=df['species'])
plt.title("Line Chart of Sepal Length vs Sepal Width")
plt.show()

#Bar-chart
plt.figure(figsize=(6,4))
sns.barplot(x=df['species'], y=df['sepal length (cm)'], color='brown')
plt.title("Bar Chart of mean Sepal Length by Species")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.iloc[:,:4].corr(), annot=True, cmap='coolwarm')
plt.title("Heatmap of feature correlation")
plt.show()
plt.figure(figsize=(8,6))
plt.scatter(df['sepal length (cm)'], df['sepal width (cm)'],s=df['petal length (cm)']*20,alpha=0.5,c='blue')
plt.show()
sns.pairplot(df.iloc[:,:4])
plt.suptitle("Pair plot of Iris Featurea",y=1.02)
plt.show()
```
