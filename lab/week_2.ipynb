```python
import pandas as pd
import numpy as np
#Data cleaning and handling missing values
df=pd.DataFrame({'Age':[25,30,np.nan,40,35],'Department':['HR','Finance','Finance',np.nan,'IT']})
print(df)
df['Age']=df['Age'].fillna(df['Age'].mean())
df['Department']=df['Department'].fillna(df['Department'].mode()[0])
print(df)
```

```python
import pandas as pd
df = pd.DataFrame({
    'ID':[1,2,2,3,3,4],
    'Name':['Alice','Bob','Bob','David','David','Charlie'],
    'Age':[25,30,30,40,40,35],
})
print("Original Dataset:")
print(df)
df_exact = df.drop_duplicates()
print("After dropping duplicates:")
df_exact
```

```python
import pandas as pd
df = pd.DataFrame({
    'ID':[1,2,2,3,3,4],
    'Name':['Alice','Bob','Bob','David','David','Charlie'],
    'Age':[25,30,30,40,40,35],
})
print("Original Dataset:")
print(df)
df_subset_id = df.drop_duplicates(subset=['ID'])
print("After dropping duplicates of id:")
print(df_subset_id)
df_subset_name = df.drop_duplicates(subset=['Name'])
print("After dropping duplicates of name:")
print(df_subset_name)

```

```python
import pandas as pd
df = pd.DataFrame({
    'Date':['2006-03-14','14/03/2006','March 14 2006','2006.14.03']
})
print("Original Dataset:")
print(df)
df['Date']=pd.to_datetime(df['Date'],errors='coerce').dt.strftime('%y-%m-%d')
print(df)
```

```python
#Case normalization (lower/upper)
df=pd.DataFrame({'Name':['Alice','Bob','Charlie','David']})
df['Name_lower']=df['Name'].str.lower()
df['Name_upper']=df['Name'].str.upper()
print(df)
```
